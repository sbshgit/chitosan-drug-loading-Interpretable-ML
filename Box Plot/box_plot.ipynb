```python
import pandas as pd
import os

# Load the Excel file
file_path = 'nested_cross_validation_results/15_feat_DT.xlsx'
df = pd.read_excel(file_path)

# Function to clean list elements and convert them to floats
def clean_and_convert(value):
    # Remove brackets, newlines, and extra spaces, then split by space or comma
    cleaned_value = value.replace('[', '').replace(']', '').replace('\n', ' ').strip()
    # Split by any whitespace or commas
    cleaned_value = cleaned_value.split()
    # Convert each element to float
    return [float(item) for item in cleaned_value]

# Apply the cleaning and conversion to both columns
df['Experimental_Drug_Loading'] = df['Experimental_Drug_Loading'].apply(clean_and_convert)
df['Predicted_Drug_Loading'] = df['Predicted_Drug_Loading'].apply(clean_and_convert)

# Calculate the absolute errors for each corresponding list element
df['DT_AE'] = df.apply(lambda row: [abs(e - p) for e, p in zip(row['Experimental_Drug_Loading'], row['Predicted_Drug_Loading'])], axis=1)

# Explode the 'DT_AE' column so that each list element is placed in its own row
df_exploded = df.explode('DT_AE')

# Define the output folder and file path
output_folder = 'AE_Box_Plot'
output_file = os.path.join(output_folder, 'AE_DT_box_plot.xlsx')

# Create the folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

# Save the flattened dataframe with absolute errors to an Excel file
df_exploded[['DT_AE']].to_excel(output_file, index=False)
```

```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import os

# Load the Excel file
file_path = 'AE_Box_Plot/Box_Plot.xlsx'
df = pd.read_excel(file_path)
# Sort columns by mean absolute error
sorted_index = df.mean().sort_values().index
df_sorted = df[sorted_index]
# Plotting the box plot
plt.figure(figsize=(12, 8))
palette = sns.color_palette("pastel")
sns.set_style("white")

boxplot = sns.boxplot(data=df_sorted, palette=palette, saturation=0.8,
                      boxprops=dict(linewidth=1.0, edgecolor='black', alpha=0.8),
                      whiskerprops=dict(linewidth=1.0, color='black'),
                      capprops=dict(linewidth=1.0, color='black'),
                      flierprops=dict(marker="d", markerfacecolor="black", markeredgecolor="black",
                                      markersize=0.5, alpha=0.2),
                      medianprops=dict(color="black", linewidth=1.0, linestyle='--'),
                      showmeans=True,
                      meanprops=dict(marker="o", markerfacecolor="black", alpha=1.0,
                                     markeredgecolor="black", markersize=4, linewidth=0.05, zorder=10))

# Add each datapoint
sns.stripplot(data=df_sorted, marker="o", edgecolor='white', alpha=0.3, size=1.5, linewidth=0.3, color='black', zorder=0)

# Titles and labels
boxplot.set_xlabel("Model index", fontsize=16, color='black', weight="bold")
boxplot.set_ylabel("Absolute error (AE)", fontsize=16, color='black', weight="bold")

# y-axis limits and interval
boxplot.set(ylim=(-0.02, 2.02), yticks=np.arange(0,135,5))

# x-axis rotation and text color
boxplot.set_xticklabels(boxplot.get_xticklabels(), rotation=0, color='black', fontsize=12, weight="bold")

# x-axis and y-axis tick color
boxplot.tick_params(colors='black', which='both')  # 'both' refers to minor and major axes
boxplot.tick_params(bottom=False, left=True)
plt.yticks(fontsize=12)
plt.xticks(fontsize=12, weight="bold")
plt.tight_layout()
# Base filename
output_base = 'ArticleFigures/BoxPLot'
# Save as PNG
plt.savefig(
    output_base + '.png',
    dpi=600,
    format='png',
    transparent=False,
    bbox_inches='tight'
)

# Save as JPG
plt.savefig(
    output_base + '.jpg',
    dpi=600,
    format='jpg',
    transparent=False,
    bbox_inches='tight'
)

# Save as JPEG
plt.savefig(
    output_base + '.jpeg',
    dpi=600,
    format='jpeg',
    transparent=False,
    bbox_inches='tight'
)

# Save as TIFF
plt.savefig(
    output_base + '.tiff',
    dpi=600,
    format='tiff',
    transparent=False,
    bbox_inches='tight'
)

# Save as PDF
plt.savefig(
    output_base + '.pdf',
    dpi=600,
    format='pdf',
    transparent=False,
    bbox_inches='tight'
)

plt.show()
```
