```python
import pandas as pd
import pickle
from sklearn.preprocessing import StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
import shap
import numpy as np
import pandas as pd
```

```python

datafile = "Lab_Formulations_Force_plot.xlsx"
pdatafile = "Dataset_14_features.xlsx"

df = pd.read_excel(datafile)
pdf = pd.read_excel(pdatafile)
with open('14_feat_NGB_model.pkl', 'rb') as file:
    NGB_model = pickle.load(file)

X = df.drop(['Formulation_number', 'Modification',
            'Drug', 'Cross_linker', 'DL'], axis='columns')
pX = pdf.drop(['Formulation_number', 'Modification',
              'Drug', 'Cross_linker', 'DL'], axis='columns')
y = df['DL']
```

```python
stdScale = StandardScaler().fit(pX)
X_scale = stdScale.transform(pX)
```

```python
NGB_model
```

```python
bestModel = NGB_model
bestModel.fit(X, y)
explainer = shap.TreeExplainer(NGB_model)
shap_values = explainer(X)
```

# Chitosan-Betamethasone

```python
shap.plots.force(
    explainer.expected_value,
    shap_values.values[0],
    feature_names=X.columns.tolist(),
    link='identity',
    plot_cmap='RdBu',
    matplotlib=True,
    show=False,
    ordering_keys=None,
    text_rotation=0,
    contribution_threshold=0.05
)
plt.subplots_adjust(left=0.01, right=0.99, top=0.92, bottom=0.22)
plt.savefig(
    'ArticleFigures/Figure_10A.png',
    dpi=600,
    format='png',
    bbox_inches='tight',
)
plt.savefig(
    'ArticleFigures/Figure_10A.jpg',
    dpi=600,
    format='jpg',
    bbox_inches='tight',
)
plt.savefig(
    'ArticleFigures/Figure_10A.jpeg',
    dpi=600,
    format='jpeg',
    bbox_inches='tight',
)
plt.savefig(
    'ArticleFigures/Figure_10A.tiff',
    dpi=600,
    format='tiff',
    bbox_inches='tight',
)
plt.savefig(
    'ArticleFigures/Figure_10A.pdf',
    dpi=600,
    format='pdf',
    bbox_inches='tight',
)
plt.show()
```
