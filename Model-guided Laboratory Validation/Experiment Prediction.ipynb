# Betamethasone -loaded CsNPs

```python
import pandas as pd
import numpy as np
import pickle
from sklearn.preprocessing import StandardScaler


# ============================================================
# 1. FILE PATHS
# ============================================================

model_path = "NGB_14_features_model.pkl"
training_path = "Dataset_14_features.xlsx"
lab_path = "Lab Formulations.xlsx"


# ============================================================
# 2. LOAD MODEL
# ============================================================

with open(model_path, "rb") as file:
    NGB_model = pickle.load(file)

print("Loaded model type:")
print(type(NGB_model))

if not hasattr(NGB_model, "predict"):
    raise TypeError(
        "The loaded object is not a fitted model. "
        f"Loaded object type: {type(NGB_model)}"
    )


# ============================================================
# 3. LOAD DATA
# ============================================================

df_train = pd.read_excel(training_path)
lab_df = pd.read_excel(lab_path)


# ============================================================
# 4. DEFINE THE 14 MODEL INPUT FEATURES
# ============================================================

FEATURES = [
    "Cross/C ratio",
    "D/C ratio",
    "Chitosan_MW",
    "Chitosan_DDA",
    "DS",
    "Modifier_LogP",
    "Modifier_pka",
    "Cross_MW",
    "Cross_pka",
    "Drug_MW",
    "Drug_pka",
    "PS",
    "NumHDonors",
    "MolLogP"
]


# ============================================================
# 5. CHECK REQUIRED COLUMNS
# ============================================================

missing_train = [
    col for col in FEATURES
    if col not in df_train.columns
]

missing_lab = [
    col for col in FEATURES
    if col not in lab_df.columns
]

if missing_train:
    raise ValueError(
        f"Training dataset is missing: {missing_train}"
    )

if missing_lab:
    raise ValueError(
        f"Lab formulation file is missing: {missing_lab}"
    )


# ============================================================
# 6. FIT SCALER USING TRAINING DATA
# ============================================================

X_train = df_train[FEATURES].astype(float)

scaler = StandardScaler()
scaler.fit(X_train)


# ============================================================
# 7. SELECT FORMULATION 1 ONLY
# ============================================================

formulation_1 = lab_df[
    lab_df["Formulation_number"] == 1
].copy()

if formulation_1.empty:
    raise ValueError("Formulation_number 1 was not found in the lab file.")

if len(formulation_1) > 1:
    raise ValueError(
        "More than one row was found for Formulation_number 1."
    )


# ============================================================
# 8. PREPARE FORMULATION 1 INPUTS
# ============================================================

X_lab_1 = formulation_1[FEATURES].astype(float)

X_lab_1_scaled = scaler.transform(X_lab_1)


# ============================================================
# 9. PREDICT DL FOR FORMULATION 1
# ============================================================

predicted_DL_1 = NGB_model.predict(X_lab_1_scaled)[0]


# ============================================================
# 10. PRINT RESULT
# ============================================================

row = formulation_1.iloc[0]

print("\n" + "=" * 80)
print("NGB PREDICTION OF DRUG LOADING")
print("=" * 80)

print(f"Formulation: {int(row['Formulation_number'])}")
print(f"Modification: {row['Modification']}")
print(f"Drug: {row['Drug']}")
print(f"Predicted DL = {predicted_DL_1:.2f}%")

print("=" * 80)
```
