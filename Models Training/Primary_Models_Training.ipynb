```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error


# ============================================================
# 1. MODEL FILES
# ============================================================

files = {
    "NGB":   "nested_cross_validation_results/15_feat_NGB.xlsx",
    "XGB":   "nested_cross_validation_results/15_feat_XGB.xlsx",
    "RF":    "nested_cross_validation_results/15_feat_RF.xlsx",
    "k-NN":  "nested_cross_validation_results/15_feat_kNN.xlsx",
    "LGBM":  "nested_cross_validation_results/15_feat_LGBM.xlsx",
    "DT":    "nested_cross_validation_results/15_feat_DT.xlsx",
    "SVR":   "nested_cross_validation_results/15_feat_SVR.xlsx",
    "Lasso": "nested_cross_validation_results/15_feat_lasso.xlsx",
    "PLS":   "nested_cross_validation_results/15_feat_PLS.xlsx",
    "MLR":   "nested_cross_validation_results/15_feat_MLR.xlsx"
}


# ============================================================
# 2. FUNCTION TO PARSE STORED ARRAYS
# ============================================================

def parse_array(value):

    if isinstance(value, np.ndarray):
        return value.astype(float)

    value = str(value).strip()

    # Remove brackets
    value = value.strip("[]")

    # Convert whitespace-separated numbers
    return np.fromstring(value, sep=" ")


# ============================================================
# 3. CALCULATE BASELINE AND IMPROVEMENT
# ============================================================

results = []

all_model_y_true = {}

for model_name, file_path in files.items():

    print(f"Processing {model_name}...")

    df = pd.read_excel(file_path)

    all_y_true = []
    all_y_pred = []

    # --------------------------------------------------------
    # Collect all outer-test observations and predictions
    # --------------------------------------------------------

    for _, row in df.iterrows():

        y_true = parse_array(
            row["Experimental_Drug_Loading"]
        )

        y_pred = parse_array(
            row["Predicted_Drug_Loading"]
        )

        if len(y_true) != len(y_pred):
            raise ValueError(
                f"{model_name}: length mismatch in "
                f"iteration {row['Iter']}"
            )

        all_y_true.extend(y_true)
        all_y_pred.extend(y_pred)

    all_y_true = np.asarray(all_y_true, dtype=float)
    all_y_pred = np.asarray(all_y_pred, dtype=float)

    all_model_y_true[model_name] = all_y_true

    # --------------------------------------------------------
    # Model outer-test metrics
    # --------------------------------------------------------

    model_mae = mean_absolute_error(
        all_y_true,
        all_y_pred
    )

    model_rmse = np.sqrt(
        mean_squared_error(
            all_y_true,
            all_y_pred
        )
    )

    # --------------------------------------------------------
    # Naïve global-mean predictor
    # --------------------------------------------------------

    global_mean = np.mean(all_y_true)

    baseline_predictions = np.full(
        len(all_y_true),
        global_mean,
        dtype=float
    )

    baseline_mae = mean_absolute_error(
        all_y_true,
        baseline_predictions
    )

    baseline_rmse = np.sqrt(
        mean_squared_error(
            all_y_true,
            baseline_predictions
        )
    )

    # --------------------------------------------------------
    # Improvement over baseline
    # --------------------------------------------------------

    mae_improvement = (
        (baseline_mae - model_mae)
        / baseline_mae
    ) * 100

    rmse_improvement = (
        (baseline_rmse - model_rmse)
        / baseline_rmse
    ) * 100

    results.append({
        "Model": model_name,
        "Baseline MAE": baseline_mae,
        "Baseline RMSE": baseline_rmse,
        "Model Test MAE": model_mae,
        "Model Test RMSE": model_rmse,
        "MAE improvement (%)": mae_improvement,
        "RMSE improvement (%)": rmse_improvement
    })


# ============================================================
# 4. CREATE RESULTS TABLE
# ============================================================

results_df = pd.DataFrame(results)

model_order = [
    "NGB",
    "XGB",
    "RF",
    "k-NN",
    "LGBM",
    "DT",
    "SVR",
    "Lasso",
    "PLS",
    "MLR"
]

results_df["Model"] = pd.Categorical(
    results_df["Model"],
    categories=model_order,
    ordered=True
)

results_df = results_df.sort_values(
    "Model"
).reset_index(drop=True)


# ============================================================
# 5. PRINT RESULTS
# ============================================================

print("\n")
print("=" * 110)
print("NAÏVE MEAN BASELINE AND IMPROVEMENT")
print("=" * 110)

print(
    results_df[
        [
            "Model",
            "Baseline MAE",
            "Baseline RMSE",
            "Model Test MAE",
            "Model Test RMSE",
            "MAE improvement (%)",
            "RMSE improvement (%)"
        ]
    ].to_string(
        index=False,
        float_format=lambda x: f"{x:.3f}"
    )
)

print("=" * 110)


# ============================================================
# 6. SAVE RESULTS
# ============================================================

output_file = (
    "nested_cross_validation_results/"
    "15_feature_baseline_improvement.xlsx"
)

results_df.to_excel(
    output_file,
    index=False
)

print(
    f"\nResults saved to:\n{output_file}"
)


# ============================================================
# 7. CHECK WHETHER BASELINE IS IDENTICAL ACROSS MODELS
# ============================================================

print("\nBaseline consistency check:")

print(
    "Baseline MAE values:",
    results_df["Baseline MAE"].round(6).tolist()
)

print(
    "Baseline RMSE values:",
    results_df["Baseline RMSE"].round(6).tolist()
)
```

```python
import os
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 1. Paths and model files
# ============================================================

files = {
    "NGB":   "nested_cross_validation_results/15_feat_NGB.xlsx",
    "XGB":   "nested_cross_validation_results/15_feat_XGB.xlsx",
    "RF":    "nested_cross_validation_results/15_feat_RF.xlsx",
    "k-NN":  "nested_cross_validation_results/15_feat_kNN.xlsx",
    "LGBM":  "nested_cross_validation_results/15_feat_LGBM.xlsx",
    "DT":    "nested_cross_validation_results/15_feat_DT.xlsx",
    "SVR":   "nested_cross_validation_results/15_feat_SVR.xlsx",
    "Lasso": "nested_cross_validation_results/15_feat_lasso.xlsx",
    "PLS":   "nested_cross_validation_results/15_feat_PLS.xlsx",
    "MLR":   "nested_cross_validation_results/15_feat_MLR.xlsx"
}


# ============================================================
# 2. Function to parse arrays stored in Excel
# ============================================================

def parse_array(value):
    """
    Convert an Excel cell such as:
    [14.9   0.54   7.88 15.9 ...]
    into a NumPy array.
    """

    if isinstance(value, np.ndarray):
        return value.astype(float)

    value = str(value).strip()
    value = value.strip("[]")

    return np.fromstring(value, sep=" ")


# ============================================================
# 3. Calculate metrics for each model
# ============================================================

results = []

for model_name, file_path in files.items():

    print(f"Processing {model_name}...")

    df = pd.read_excel(file_path)

    # --------------------------------------------------------
    # Check required columns
    # --------------------------------------------------------

    required_columns = [
        "Valid Score",
        "Test Score",
        "Experimental_Drug_Loading",
        "Predicted_Drug_Loading"
    ]

    missing = [
        col for col in required_columns
        if col not in df.columns
    ]

    if missing:
        raise ValueError(
            f"{model_name}: missing columns: {missing}"
        )

    # --------------------------------------------------------
    # Validation MAE
    # --------------------------------------------------------

    validation_mae_mean = df["Valid Score"].mean()
    validation_mae_sd = df["Valid Score"].std(ddof=1)

    # --------------------------------------------------------
    # Test MAE
    # --------------------------------------------------------

    test_mae_mean = df["Test Score"].mean()
    test_mae_sd = df["Test Score"].std(ddof=1)

    # --------------------------------------------------------
    # Calculate outer-fold RMSE and R²
    # --------------------------------------------------------

    rmse_scores = []
    r2_scores = []

    all_y_true = []
    all_y_pred = []

    for _, row in df.iterrows():

        y_true = parse_array(
            row["Experimental_Drug_Loading"]
        )

        y_pred = parse_array(
            row["Predicted_Drug_Loading"]
        )

        if len(y_true) != len(y_pred):
            raise ValueError(
                f"{model_name}: mismatch in Iter {row['Iter']} "
                f"(true={len(y_true)}, pred={len(y_pred)})"
            )

        # Fold-level metrics
        rmse = np.sqrt(
            mean_squared_error(
                y_true,
                y_pred
            )
        )

        r2 = r2_score(y_true, y_pred)
        

        rmse_scores.append(rmse)
        r2_scores.append(r2)

        # For global baseline
        all_y_true.extend(y_true)
        all_y_pred.extend(y_pred)

    rmse_mean = np.mean(rmse_scores)
    rmse_sd = np.std(rmse_scores, ddof=1)

    r2_mean = np.nanmean(r2_scores)
    
    # ========================================================
    # 4. Global naïve mean baseline
    # ========================================================

    all_y_true = np.asarray(all_y_true, dtype=float)
    all_y_pred = np.asarray(all_y_pred, dtype=float)

    global_mean_dl = np.mean(all_y_true)

    baseline_predictions = np.full(
        len(all_y_true),
        global_mean_dl,
        dtype=float
    )

    baseline_mae = mean_absolute_error(
        all_y_true,
        baseline_predictions
    )

    baseline_rmse = np.sqrt(
        mean_squared_error(
            all_y_true,
            baseline_predictions
        )
    )

    # ========================================================
    # 5. Improvement over naïve baseline
    # ========================================================

    mae_improvement = (
        (baseline_mae - test_mae_mean)
        / baseline_mae
    ) * 100

    rmse_improvement = (
        (baseline_rmse - rmse_mean)
        / baseline_rmse
    ) * 100

    # ========================================================
    # 6. Store results
    # ========================================================

    results.append({
        "Model": model_name,
        "Features": 15,

        "Validation MAE": validation_mae_mean,
        "Validation MAE SD": validation_mae_sd,

        "Test MAE": test_mae_mean,
        "Test MAE SD": test_mae_sd,

        "Test RMSE": rmse_mean,

        "Test R²": r2_mean,

        "Baseline MAE": baseline_mae,
        "Baseline RMSE": baseline_rmse,

        "MAE improvement (%)": mae_improvement,
        "RMSE improvement (%)": rmse_improvement,

        "Global mean DL": global_mean_dl
    })


# ============================================================
# 7. Create results DataFrame
# ============================================================

results_df = pd.DataFrame(results)


# ============================================================
# 8. Sort by Test MAE: lowest first
# ============================================================

results_df.sort_values(
    by="Test MAE",
    ascending=True,
    inplace=True
)

results_df.reset_index(
    drop=True,
    inplace=True
)


# ============================================================
# 9. Round values for presentation
# ============================================================

results_display = results_df.copy()

numeric_columns = [
    "Validation MAE",
    "Validation MAE SD",
    "Test MAE",
    "Test MAE SD",
    "Test RMSE",
    "Test R²",
    "Baseline MAE",
    "Baseline RMSE",
    "MAE improvement (%)",
    "RMSE improvement (%)",
    "Global mean DL"
]

results_display[numeric_columns] = (
    results_display[numeric_columns].round(3)
)


# ============================================================
# 10. Print final table
# ============================================================

print("\n")
print("=" * 140)
print("15-FEATURE ALL-MODEL PERFORMANCE")
print("=" * 140)

print(
    results_display.to_string(index=False)
)

print("=" * 140)


# ============================================================
# 11. Save Excel file
# ============================================================

output_path = (
    "nested_cross_validation_results/"
    "15_feature_all_models_metrics.xlsx"
)

os.makedirs(
    "nested_cross_validation_results",
    exist_ok=True
)

results_display.to_excel(
    output_path,
    index=False
)

print(
    f"\nResults saved to:\n{output_path}"
)
```
