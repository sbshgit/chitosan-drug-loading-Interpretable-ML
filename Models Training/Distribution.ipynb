Distribution of Drug Loading

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_excel("Dataset_15_features.xlsx")
f, boxplot = plt.subplots(figsize=(10,6))
ax = sns.histplot(df['DL'], binwidth=1, alpha=0.85, color='palevioletred',  kde=True)
ax.lines[0].set_color('black')
plt.ylabel("Count", fontsize=14, color='black',  weight="bold")
plt.xlabel("Drug Loading", fontsize=14, color='black', weight="bold")
plt.xticks(np.arange(0, 101, 5)) 
plt.yticks(np.arange(0,12))
plt.yticks(fontsize=10)
plt.xticks(fontsize=10)
plt.tight_layout()
output_base = 'ArticleFigures/HistPlot1'
# Save as PNG
plt.savefig(
    output_base + '.png',
    dpi=600,
    format='png',
    transparent=False,
    bbox_inches='tight'
)

# Save as JPG
plt.savefig(
    output_base + '.jpg',
    dpi=600,
    format='jpg',
    transparent=False,
    bbox_inches='tight'
)

# Save as JPEG
plt.savefig(
    output_base + '.jpeg',
    dpi=600,
    format='jpeg',
    transparent=False,
    bbox_inches='tight'
)

# Save as TIFF
plt.savefig(
    output_base + '.tiff',
    dpi=600,
    format='tiff',
    transparent=False,
    bbox_inches='tight'
)

# Save as PDF
plt.savefig(
    output_base + '.pdf',
    dpi=600,
    format='pdf',
    transparent=False,
    bbox_inches='tight'
)
plt.show()
```

Distribution of Chitosan Modifications 

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_excel("Dataset_15_features.xlsx")
f, boxplot = plt.subplots(figsize=(10,6))
ax = sns.histplot(df["Modification"], binwidth=1, alpha= 0.9 , color='orange', kde=True)
ax.lines[0].set_color('black')
plt.ylabel("Count", fontsize=14, color='black',  weight="bold")
plt.xlabel("Modification", fontsize=14, color='black', weight="bold")
plt.yticks(np.arange(0,120,10))
plt.yticks(fontsize=10)
plt.xticks(rotation=90, fontsize=10)
plt.tight_layout()
output_base = 'ArticleFigures/HistPlot2'
# Save as PNG
plt.savefig(
    output_base + '.png',
    dpi=600,
    format='png',
    transparent=False,
    bbox_inches='tight'
)

# Save as JPG
plt.savefig(
    output_base + '.jpg',
    dpi=600,
    format='jpg',
    transparent=False,
    bbox_inches='tight'
)

# Save as JPEG
plt.savefig(
    output_base + '.jpeg',
    dpi=600,
    format='jpeg',
    transparent=False,
    bbox_inches='tight'
)

# Save as TIFF
plt.savefig(
    output_base + '.tiff',
    dpi=600,
    format='tiff',
    transparent=False,
    bbox_inches='tight'
)

# Save as PDF
plt.savefig(
    output_base + '.pdf',
    dpi=600,
    format='pdf',
    transparent=False,
    bbox_inches='tight'
)
plt.show()
```
